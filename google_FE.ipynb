```python
# === Пути ===

SAMPLE_PATH = Path("FE_google/df_sample_google.parquet")

# Папка, где лежат файлы истории по VM.
# Например:
# FE_google/filtered_sorted/a_290470320252_441.parquet
HISTORY_DIR = Path("FE_google/filtered_sorted.parquet")

# Куда сохранить результат
OUTPUT_DIR = Path("FE_google/features_result")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_PATH = OUTPUT_DIR / "features_sample_google.parquet"


# === Параметры временного ряда ===

STEP_MINUTES = 5

POINTS_PER_DAY = 24 * 60 // STEP_MINUTES  # 288

LAG_5M = 1
LAG_10M = 2
LAG_15M = 3
LAG_20M = 4

DAY_1 = POINTS_PER_DAY       # 288
DAY_2 = 2 * POINTS_PER_DAY   # 576
DAY_3 = 3 * POINTS_PER_DAY   # 864
DAY_4 = 4 * POINTS_PER_DAY   # 1152

TARGET_24H_SHIFT = POINTS_PER_DAY  # +288 точек


# === Параллелизм ===

N_WORKERS = 6

# Сколько результатов по VM накапливать перед записью на диск
WRITE_BATCH_SIZE = 132
```

```python
def get_vm_history_path(vm_id: str) -> Path | None:

    partition_dir = HISTORY_DIR / f"vm_id={vm_id}"

    if not partition_dir.exists():

        return None

    parquet_files = list(partition_dir.glob("*.parquet"))

    if not parquet_files:

        return None

    return parquet_files[0]
```

```python
def prefix_mean(csum: np.ndarray, left_idx: np.ndarray, right_idx: np.ndarray) -> np.ndarray:
    """
    Считает mean(x[left:right]) через prefix sum.

    csum имеет длину n + 1:
    csum[0] = 0
    csum[k] = sum(x[:k])
    """

    result = np.full(len(left_idx), np.nan, dtype=np.float32)

    valid = (
        (left_idx >= 0) &
        (right_idx >= 0) &
        (right_idx > left_idx) &
        (right_idx < len(csum)) &
        (left_idx < len(csum))
    )

    if valid.any():
        window_sum = csum[right_idx[valid]] - csum[left_idx[valid]]
        window_len = right_idx[valid] - left_idx[valid]
        result[valid] = (window_sum / window_len).astype(np.float32)

    return result
```

```python
def prefix_std_before(
    csum: np.ndarray,
    csum2: np.ndarray,
    idx: np.ndarray
) -> np.ndarray:
    """
    Считает std(x[:i]) для каждого i через prefix sums.

    Используем population std:
    var = E[x^2] - E[x]^2
    """

    result = np.full(len(idx), np.nan, dtype=np.float32)

    valid = idx > 1

    if valid.any():
        i = idx[valid]

        mean = csum[i] / i
        mean2 = csum2[i] / i

        var = mean2 - mean ** 2
        var = np.maximum(var, 0.0)

        result[valid] = np.sqrt(var).astype(np.float32)

    return result
```

```python
def autocorr_lag1_before(x: np.ndarray, idx: np.ndarray) -> np.ndarray:
    """
    Считает autocorr lag-1 для истории до каждой точки i.

    Для точки i используются пары:
    a = x[0 : i - 1]
    b = x[1 : i]

    Это автокорреляция с лагом 5 минут.
    """

    n = len(x)
    result = np.full(len(idx), np.nan, dtype=np.float32)

    if n < 3:
        return result

    x = x.astype(np.float64)

    # Для пар x[k], x[k+1]
    a = x[:-1]
    b = x[1:]

    # Prefix sums по a, b, a^2, b^2, a*b
    c_a = np.concatenate([[0.0], np.cumsum(a)])
    c_b = np.concatenate([[0.0], np.cumsum(b)])
    c_a2 = np.concatenate([[0.0], np.cumsum(a * a)])
    c_b2 = np.concatenate([[0.0], np.cumsum(b * b)])
    c_ab = np.concatenate([[0.0], np.cumsum(a * b)])

    # Для истории до i число пар = i - 1
    valid = idx > 2

    if valid.any():
        i = idx[valid]
        m = i - 1

        sum_a = c_a[m]
        sum_b = c_b[m]
        sum_a2 = c_a2[m]
        sum_b2 = c_b2[m]
        sum_ab = c_ab[m]

        mean_a = sum_a / m
        mean_b = sum_b / m

        cov = (sum_ab / m) - mean_a * mean_b

        var_a = (sum_a2 / m) - mean_a ** 2
        var_b = (sum_b2 / m) - mean_b ** 2

        denom = np.sqrt(np.maximum(var_a, 0.0) * np.maximum(var_b, 0.0))

        corr = np.full(len(m), np.nan, dtype=np.float64)
        ok = denom > 0
        corr[ok] = cov[ok] / denom[ok]

        result[valid] = corr.astype(np.float32)

    return result
```

```python
def process_one_vm(vm_id: str) -> pd.DataFrame:
    """
    Обрабатывает одну VM:

    1. Берет sample-строки этой VM.
    2. Находит parquet-файл истории.
    3. Читает историю.
    4. Находит индексы sample trace_minute в истории.
    5. Считает фичи и targets.
    6. Возвращает dataframe с результатами.
    """

    sample_part = sample_by_vm[vm_id].copy()

    history_path = get_vm_history_path(vm_id)

    if history_path is None:
        sample_part["feature_status"] = "history_file_not_found"
        return sample_part

    try:
        # Читаем только нужные колонки истории
        hist = pd.read_parquet(
            history_path,
            columns=["trace_minute", "average_cpu"]
        )
    except Exception as e:
        sample_part["feature_status"] = f"history_read_error: {str(e)[:120]}"
        return sample_part

    if hist.empty:
        sample_part["feature_status"] = "empty_history"
        return sample_part

    # На всякий случай сортируем.
    # Если ты на 100% уверен, что все уже отсортировано, эту строку можно убрать.
    hist = hist.sort_values("trace_minute", kind="mergesort").reset_index(drop=True)

    trace = hist["trace_minute"].to_numpy(dtype=np.float64)
    x = hist["average_cpu"].to_numpy(dtype=np.float64)

    n = len(x)

    # Prefix sums.
    # csum длины n+1, чтобы сумма x[a:b] = csum[b] - csum[a]
    csum = np.concatenate([[0.0], np.cumsum(x)])
    csum2 = np.concatenate([[0.0], np.cumsum(x * x)])

    sample_trace = sample_part["trace_minute"].to_numpy(dtype=np.float64)

    # Быстрый индекс через формулу:
    # idx = (sample_trace - first_trace) / 5
    first_trace = trace[0]
    raw_idx = (sample_trace - first_trace) / STEP_MINUTES
    idx = np.rint(raw_idx).astype(np.int64)

    # Проверка валидности индекса
    valid_idx = (
        (idx >= 0) &
        (idx < n) &
        np.isclose(trace[idx.clip(0, n - 1)], sample_trace)
    )

    # Если формула не сработала для части строк, fallback через searchsorted
    if not valid_idx.all():
        ss_idx = np.searchsorted(trace, sample_trace)

        ss_valid = (
            (ss_idx >= 0) &
            (ss_idx < n) &
            np.isclose(trace[ss_idx.clip(0, n - 1)], sample_trace)
        )

        idx = np.where(ss_valid, ss_idx, idx)
        valid_idx = ss_valid

    m = len(sample_part)

    # Создаем результат
    result = sample_part.copy()

    # Служебные поля
    result["history_len"] = n
    result["history_idx"] = idx
    result["feature_status"] = np.where(valid_idx, "ok", "trace_minute_not_found")

    # Инициализируем фичи NaN
    feature_cols = [
        "avg_cpu_mean_0_1d",
        "avg_cpu_mean_1_2d",
        "avg_cpu_mean_2_3d",
        "avg_cpu_mean_3_4d",
        "avg_cpu_lag_5m",
        "avg_cpu_lag_10m",
        "avg_cpu_lag_15m",
        "avg_cpu_lag_20m",
        "avg_cpu_autocorr_lag1_before",
        "avg_cpu_history_mean_before",
        "avg_cpu_history_std_before",
        "target_avg_cpu_5m",
        "target_avg_cpu_24h",
        "target_avg_cpu_all",
    ]

    for col in feature_cols:
        result[col] = np.nan

    if not valid_idx.any():
        return result

    # Для невалидных строк idx может быть мусорным,
    # поэтому дальше считаем только valid-позиции.
    valid_positions = np.where(valid_idx)[0]
    idx_valid = idx[valid_positions]

    # ---------- Лаги ----------
    def safe_lag(lag: int) -> np.ndarray:
        out = np.full(m, np.nan, dtype=np.float32)
        source_idx = idx_valid - lag
        ok = source_idx >= 0
        out[valid_positions[ok]] = x[source_idx[ok]].astype(np.float32)
        return out

    result["avg_cpu_lag_5m"] = safe_lag(LAG_5M)
    result["avg_cpu_lag_10m"] = safe_lag(LAG_10M)
    result["avg_cpu_lag_15m"] = safe_lag(LAG_15M)
    result["avg_cpu_lag_20m"] = safe_lag(LAG_20M)

    # ---------- Оконные средние ----------
    full_idx = idx.copy()

    # Для невалидных строк поставим -1, чтобы prefix_mean вернул NaN
    full_idx[~valid_idx] = -1

    result["avg_cpu_mean_0_1d"] = prefix_mean(csum, full_idx - DAY_1, full_idx)
    result["avg_cpu_mean_1_2d"] = prefix_mean(csum, full_idx - DAY_2, full_idx - DAY_1)
    result["avg_cpu_mean_2_3d"] = prefix_mean(csum, full_idx - DAY_3, full_idx - DAY_2)
    result["avg_cpu_mean_3_4d"] = prefix_mean(csum, full_idx - DAY_4, full_idx - DAY_3)

    # ---------- Историческое среднее до наблюдения ----------
    hist_mean = np.full(m, np.nan, dtype=np.float32)
    ok_mean = valid_idx & (idx > 0)
    hist_mean[ok_mean] = (csum[idx[ok_mean]] / idx[ok_mean]).astype(np.float32)
    result["avg_cpu_history_mean_before"] = hist_mean

    # ---------- Исторический std до наблюдения ----------
    result["avg_cpu_history_std_before"] = prefix_std_before(csum, csum2, full_idx)

    # ---------- Autocorr lag-1 до наблюдения ----------
    result["avg_cpu_autocorr_lag1_before"] = autocorr_lag1_before(x, full_idx)

    # ---------- Target 5m ----------
    # По твоему определению sample-наблюдение является average_cpu через 5 минут.
    target_5m = np.full(m, np.nan, dtype=np.float32)
    target_5m[valid_positions] = x[idx_valid].astype(np.float32)
    result["target_avg_cpu_5m"] = target_5m

    # ---------- Target 24h ----------
    target_24h = np.full(m, np.nan, dtype=np.float32)
    future_idx = idx_valid + TARGET_24H_SHIFT
    ok_future = future_idx < n
    target_24h[valid_positions[ok_future]] = x[future_idx[ok_future]].astype(np.float32)
    result["target_avg_cpu_24h"] = target_24h

    # ---------- Target all ----------
    target_all = np.float32(csum[-1] / n)
    result.loc[valid_idx, "target_avg_cpu_all"] = target_all

    return result
```

```python
def write_batch_to_parquet(batch_frames, output_path: Path, writer_state: dict):
    """
    Записывает список dataframe в parquet.

    writer_state нужен, чтобы writer создавался один раз,
    а потом дописывал новые row groups.
    """

    if not batch_frames:
        return

    batch_df = pd.concat(batch_frames, ignore_index=True)

    table = pa.Table.from_pandas(batch_df, preserve_index=False)

    if writer_state.get("writer") is None:
        writer_state["writer"] = pq.ParquetWriter(
            output_path,
            table.schema,
            compression="snappy"
        )

    writer_state["writer"].write_table(table)

    del batch_df, table
    gc.collect()
```

```python
# Если итоговый файл уже существует, удалим его,
# чтобы не смешать старый и новый результат.
if OUTPUT_PATH.exists():
    OUTPUT_PATH.unlink()

writer_state = {"writer": None}

batch_frames = []
errors = []

processed_vm = 0

with ThreadPoolExecutor(max_workers=N_WORKERS) as executor:
    futures = {
        executor.submit(process_one_vm, vm_id): vm_id
        for vm_id in sample_vm_ids
    }

    for future in tqdm(as_completed(futures), total=len(futures), desc="Processing VM"):
        vm_id = futures[future]

        try:
            vm_result = future.result()
            batch_frames.append(vm_result)
            processed_vm += 1

        except Exception as e:
            errors.append({
                "vm_id": vm_id,
                "error": str(e)
            })

        # Периодически пишем результат на диск
        if len(batch_frames) >= WRITE_BATCH_SIZE:
            write_batch_to_parquet(batch_frames, OUTPUT_PATH, writer_state)
            batch_frames = []

# Записываем остаток
write_batch_to_parquet(batch_frames, OUTPUT_PATH, writer_state)
batch_frames = []

# Закрываем writer
if writer_state["writer"] is not None:
    writer_state["writer"].close()

print(f"Готово.")
print(f"Обработано VM: {processed_vm:,}")
print(f"Ошибок: {len(errors):,}")
print(f"Результат сохранен в: {OUTPUT_PATH}")
```

```python
import numpy as np
import pandas as pd
from pathlib import Path
from tqdm.auto import tqdm

import torch
from torch import nn
from torch.utils.data import Dataset, DataLoader

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler

HISTORY_DIR = Path("FE_google/filtered_sorted.parquet")

TRAIN_PATH = Path("FE_google/features_result/train_df_google.parquet")
TEST_PATH = Path("FE_google/features_result/test_df_google.parquet")

SEQ_LEN = 1000
HORIZON_24H = 288

RANDOM_STATE = 42

# Чтобы быстрее обучить первую LSTM
MAX_TRAIN_OBS = 100_000
MAX_TEST_OBS = 50_000

BATCH_SIZE = 256
EPOCHS = 5
LR = 1e-3

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DEVICE
```

```python
train_df_google = pd.read_parquet(TRAIN_PATH)
test_df_google = pd.read_parquet(TEST_PATH)

train_df_google = train_df_google[train_df_google["feature_status"] == "ok"].copy()
test_df_google = test_df_google[test_df_google["feature_status"] == "ok"].copy()

train_df_google["vm_id"] = train_df_google["vm_id"].astype(str)
test_df_google["vm_id"] = test_df_google["vm_id"].astype(str)

train_df_google["trace_minute"] = train_df_google["trace_minute"].astype(float)
test_df_google["trace_minute"] = test_df_google["trace_minute"].astype(float)

print("train rows:", len(train_df_google))
print("test rows:", len(test_df_google))
print("train VM:", train_df_google["vm_id"].nunique())
print("test VM:", test_df_google["vm_id"].nunique())
```

```python
train_obs = train_df_google.sample(
    n=min(MAX_TRAIN_OBS, len(train_df_google)),
    random_state=RANDOM_STATE
).reset_index(drop=True)

test_obs = test_df_google.sample(
    n=min(MAX_TEST_OBS, len(test_df_google)),
    random_state=RANDOM_STATE
).reset_index(drop=True)

print("train obs:", len(train_obs))
print("test obs:", len(test_obs))
```

```python
def get_vm_history_path(vm_id: str) -> Path | None:
    partition_dir = HISTORY_DIR / f"vm_id={vm_id}"

    if not partition_dir.exists():
        return None

    files = list(partition_dir.glob("*.parquet"))

    if not files:
        return None

    return files[0]
```

```python
LSTM_TIME_FEATURES = [
    "average_cpu",
    "maximum_cpu",
    "maximum_memory",
]

STATIC_FEATURES = [
    "requested_cpu",
    "requested_ram",
]

def build_lstm_dataset_from_obs(obs_df: pd.DataFrame):
    X_list = []
    y_list = []
    meta_list = []

    obs_by_vm = {
        vm_id: group.reset_index(drop=True)
        for vm_id, group in obs_df.groupby("vm_id", sort=False)
    }

    for vm_id, group in tqdm(obs_by_vm.items(), total=len(obs_by_vm)):
        path = get_vm_history_path(vm_id)

        if path is None:
            continue

        hist = pd.read_parquet(
            path,
            columns=["trace_minute"] + LSTM_TIME_FEATURES
        )

        hist = hist.sort_values("trace_minute").reset_index(drop=True)

        trace = hist["trace_minute"].to_numpy(dtype=np.float64)
        values = hist[LSTM_TIME_FEATURES].to_numpy(dtype=np.float32)

        # Статические признаки берем из obs_df.
        # Они одинаковые для VM, поэтому берем первую строку группы.
        static_vals = group[STATIC_FEATURES].iloc[0].to_numpy(dtype=np.float32)

        obs_trace = group["trace_minute"].to_numpy(dtype=np.float64)

        # Ищем индекс наблюдения в истории
        first_trace = trace[0]
        idx = np.rint((obs_trace - first_trace) / 5).astype(np.int64)

        for row_pos, i in enumerate(idx):
            # По твоему условию история есть, но оставим защиту
            if i < SEQ_LEN:
                continue

            if i + HORIZON_24H >= len(values):
                continue

            # X: 1000 предыдущих точек
            x_time = values[i - SEQ_LEN : i]

            # Добавляем requested_cpu/requested_ram на каждый временной шаг
            x_static = np.repeat(static_vals.reshape(1, -1), SEQ_LEN, axis=0)

            x_window = np.concatenate([x_time, x_static], axis=1)

            # y_5m: значение в точке наблюдения
            y_5m = values[i, 0]

            # y_24h: значение через 288 точек
            y_24h = values[i + HORIZON_24H, 0]

            X_list.append(x_window)
            y_list.append([y_5m, y_24h])

            meta_list.append({
                "vm_id": vm_id,
                "trace_minute": group.loc[row_pos, "trace_minute"],
                "history_start_idx": i - SEQ_LEN,
                "history_end_idx": i - 1,
                "observation_idx": i,
                "target_24h_idx": i + HORIZON_24H,
            })

    X = np.stack(X_list).astype(np.float32)
    y = np.array(y_list, dtype=np.float32)
    meta = pd.DataFrame(meta_list)

    return X, y, meta
```

```python
X_train, y_train, meta_train = build_lstm_dataset_from_obs(train_obs)
X_test, y_test, meta_test = build_lstm_dataset_from_obs(test_obs)

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
num_features = X_train.shape[-1]

scaler_X = StandardScaler()

X_train_2d = X_train.reshape(-1, num_features)
X_test_2d = X_test.reshape(-1, num_features)

scaler_X.fit(X_train_2d)

X_train_scaled = scaler_X.transform(X_train_2d).reshape(X_train.shape).astype(np.float32)
X_test_scaled = scaler_X.transform(X_test_2d).reshape(X_test.shape).astype(np.float32)

scaler_y = StandardScaler()
scaler_y.fit(y_train)

y_train_scaled = scaler_y.transform(y_train).astype(np.float32)
y_test_scaled = scaler_y.transform(y_test).astype(np.float32)
```

```python
class LSTMDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]


train_dataset = LSTMDataset(X_train_scaled, y_train_scaled)
test_dataset = LSTMDataset(X_test_scaled, y_test_scaled)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    drop_last=False
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    drop_last=False
)
```

```python
class CPUForecastLSTM(nn.Module):
    def __init__(self, input_size, hidden_size=64, num_layers=2, dropout=0.2):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True,
            dropout=dropout
        )

        self.head = nn.Sequential(
            nn.Linear(hidden_size, 64),
            nn.ReLU(),
            nn.Linear(64, 2)
        )

    def forward(self, x):
        out, _ = self.lstm(x)

        # Берем hidden state последнего временного шага
        last_hidden = out[:, -1, :]

        preds = self.head(last_hidden)

        return preds


model = CPUForecastLSTM(
    input_size=X_train_scaled.shape[-1],
    hidden_size=64,
    num_layers=2,
    dropout=0.2
).to(DEVICE)

optimizer = torch.optim.Adam(model.parameters(), lr=LR)
criterion = nn.MSELoss()
```

```python
for epoch in range(EPOCHS):
    model.train()
    train_losses = []

    for X_batch, y_batch in tqdm(train_loader, desc=f"Epoch {epoch+1}/{EPOCHS}"):
        X_batch = X_batch.to(DEVICE)
        y_batch = y_batch.to(DEVICE)

        optimizer.zero_grad()

        preds = model(X_batch)
        loss = criterion(preds, y_batch)

        loss.backward()
        optimizer.step()

        train_losses.append(loss.item())

    print(f"Epoch {epoch+1}: train loss = {np.mean(train_losses):.6f}")
```

```python
model.eval()

preds_scaled_list = []

with torch.no_grad():
    for X_batch, _ in tqdm(test_loader, desc="Predicting"):
        X_batch = X_batch.to(DEVICE)

        preds_scaled = model(X_batch)

        preds_scaled_list.append(preds_scaled.cpu().numpy())

preds_scaled = np.vstack(preds_scaled_list)

preds = scaler_y.inverse_transform(preds_scaled)

meta_test["pred_lstm_5m"] = preds[:, 0]
meta_test["pred_lstm_24h"] = preds[:, 1]

meta_test["true_5m"] = y_test[:, 0]
meta_test["true_24h"] = y_test[:, 1]

meta_test.head()
```

```python
def regression_metrics(y_true, y_pred, name):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = mean_squared_error(y_true, y_pred, squared=False)
    r2 = r2_score(y_true, y_pred)

    return {
        "target": name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2,
    }


metrics = pd.DataFrame([
    regression_metrics(
        meta_test["true_5m"],
        meta_test["pred_lstm_5m"],
        "average_cpu_5m"
    ),
    regression_metrics(
        meta_test["true_24h"],
        meta_test["pred_lstm_24h"],
        "average_cpu_24h"
    ),
])

metrics
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

def evaluate_model(y_true, y_pred, name=""):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    # на всякий случай убираем NaN/inf
    mask = np.isfinite(y_true) & np.isfinite(y_pred)
    y_true = y_true[mask]
    y_pred = y_pred[mask]

    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)

    underestimations = (y_pred < y_true).sum()
    svr = underestimations / len(y_true) * 100

    overestimations = (y_pred > y_true).sum()
    ovr = overestimations / len(y_true) * 100

    print(f"\n{'='*50}")
    print(f"{name}")
    print(f"{'='*50}")
    print(f"MAE:  {mae:.6f}")
    print(f"RMSE: {rmse:.6f}")
    print(f"R2:   {r2:.6f}")
    print(f"SVR (недооценка): {svr:.2f}% ({underestimations}/{len(y_true)})")
    print(f"OVR (переоценка): {ovr:.2f}% ({overestimations}/{len(y_true)})")

    return {
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2,
        "SVR": svr,
        "OVR": ovr,
        "n": len(y_true),
    }
```
